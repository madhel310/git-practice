##Task 2 – Working with multiple formats <br>
You are building a quick script to check laboratory equipment calibration status. <br>
You will ead target settings from a YAML file, look up sensor locations in an Excel sheet, process calibration logs from a CSV file, and output the out-of-date sensors to a JSON file. <br>
Write a script check_sensors.py that executes the following pipeline: <br>
- Read settings:
- Read config.yml and get max_days_since_calibration and output_file
- Read and join data:
- Read sensors.xlsx
- Read calibrations.csv
- Match each sensor's location/owner with its calibration status.
- Filter and export:
- Identify sensors where days_since_calibration > max_days_since_calibration.
- Export the overdue sensors as a formatted JSON array to the filename
specified in config.yml using json.dump(..., indent=2).
- This json file should include sensor_id, lab/owner information as well as
days_since_calibration for the overdue sensors only
- Submit the script and the generated json file either as usual on Canvas, or in a github repository (must be public, submit the link to the repository)


```python
import json
import yaml
import pandas as pd

"""
Leser sensor og Kalibreringsdata, identifiserer utløpte sensorer,
og eksporterer de som JSON fil
"""

def main():
    """ Les filer, finn utløpte sensorer, og eksporter dem til JSON"""

    #Les konfigurasjonen
    with open("config.yml", "r") as file:
        config = yaml.safe_load(file)

    max_days = config["max_days_since_calibration"]
    output_file = config["output_file"]

    #Les data filer
    sensors_df = pd.read_excel("sensors.xlsx")
    calibrations_df = pd.read_csv("calibrations.csv")

    #slpr sammen sensor og kalibrerings info
    merged_df = pd.merge(
        sensors_df,
        calibrations_df,
        on="sensor_id"
    )

    #filtrerer utløpte sensorer.
    overdue_df = merged_df[
        merged_df["days_since_calibration"] > max_days
    ]

    #Kontverter til en liste med ordbøker
    result = overdue_df.to_dict(orient="records")

    #eksporter JSON
    with open(output_file, "w") as file:
        json.dump(result, file, indent = 2)

    print(f"Eksportert {len(result)} utløpte sensorer til {output_file}")

if __name__ == "__main__":
    main()


    
```
